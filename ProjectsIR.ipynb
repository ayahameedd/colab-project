**🛠️ NLP Environment Setup & Text Preprocessing**

This notebook initializes the environment for Natural Language Processing (NLP) tasks using the NLTK (Natural Language Toolkit) library. The code block above prepares the workspace by importing essential modules and downloading the necessary linguistic datasets.

**📋 Included Components:**

Tokenization (punkt): Used to split raw text into individual words or sentences.

Stopwords Removal: Downloads a list of common words (e.g., "the", "is", "at") that are often removed to focus on the core meaning of the text.

Stemming (PorterStemmer): Provides the tools to reduce words to their root form (e.g., "running" → "run").

String Handling: Includes standard Python string operations for removing punctuation.



```python
import nltk
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download necessary NLTK data packages
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

print("downloaded successfully")
```

**🔡 Step 1: Text Tokenization & Formatting**

In this section, we take the raw text and break it down into smaller, manageable pieces called Tokens. This is a fundamental step because machines process language better when it is divided into individual units (words and punctuation) rather than one long string.

**🔍 What’s happening in this code:**

**textwrap.fill():** We use this utility to ensure our output remains readable. It automatically wraps long strings of text to a maximum width (80 characters), preventing horizontal scrolling in the console.

**word_tokenize():** This is the core NLP function from nltk. Unlike a simple .split() method, this intelligent tokenizer understands how to separate punctuation from words (e.g., it knows "quickly!" should be "quickly" and "!") correctly.

```python
import textwrap

text = """Information Retrieval is the science of searching for information in documents.
Search engines use complex algorithms to retrieve relevant documents quickly!
However, building a good search engine requires proper text preprocessing.
For example, the system must tokenize sentences, remove punctuation marks like commas or periods,
and apply stemming to words like ' running', ' ran', and ' runs'."""

print("ORIGINAL TEXT ")
print(textwrap.fill(text, width=80))

print("\n" + "=" * 80 + "\n")

# Step 1: Tokenize words
tokens = word_tokenize(text)

print(" STEP 1: Tokenization")
result_step1 = ' '.join(tokens)
print(textwrap.fill(result_step1, width=80))
```

**🔡 Step 2: Case Folding (Lowercasing)**

Computers are picky—they treat "Information" and "information" as two completely different words.

What we do: Convert every token to lowercase using a list comprehension.

Why? This ensures that the frequency of a word is counted correctly regardless of its position in a sentence.

**✂️ Step 3: Punctuation Removal**

Symbols like commas, periods, and exclamation marks are essential for human reading but often act as "noise" for basic text analysis.

What we do: We filter the list to keep only the tokens that are not found in the standard string.punctuation library.

Result: We are left with purely alphanumeric content, making it much easier to analyze the actual vocabulary of the document.

```python
# Step 2: Convert to lowercase
lowercase_tokens = [word.lower() for word in tokens]

print(" STEP 2: Lowercase\n")
for i in range(0, len(lowercase_tokens), 10):
    print(' '.join(lowercase_tokens[i:i+10]))

print("\n" + "-" * 60 + "\n")

# Step 3: Remove punctuation
no_punct_tokens = [word for word in lowercase_tokens if word not in string.punctuation]

print(" STEP 3: No Punctuation\n")
for i in range(0, len(no_punct_tokens), 10):
    print(' '.join(no_punct_tokens[i:i+10]))
```

**🛑 Step 4: Stop Words Removal**

In this step, we filter out the "grammatical glue" of the language. Stop words are common words like "the", "is", "in", and "at" that appear frequently in almost every document but don't carry significant unique meaning for search or analysis.

🔍 How it works:
The Dictionary: We use NLTK's predefined list of English stop words.

The Filter: We check every token from our previous step; if it exists in the stop words list, we discard it.

Optimization: We convert the list to a set first because looking up items in a set is much faster than in a list—this is a small but important trick for handling large datasets!

```python
# Step 4: Remove stop words
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in no_punct_tokens if word not in stop_words]

print(" STEP 4: Stop Words Removed\n")


for i in range(0, len(filtered_tokens), 10):
    print(' '.join(filtered_tokens[i:i+10]))
```

**Step 5: Stemming & Final Comparison**

**✂️ What is Stemming?**

Words often appear in different forms (e.g., running, runs, ran). To a computer, these might look like different concepts. Stemming chops off the suffixes to reach the base word.

The Tool: We use the Porter Stemmer, which is the most common algorithm for English.

Example: The words "Information" and "Informative" might both be reduced to the stem "inform".

**🏁 Final Output Analysis**

The code concludes by displaying a side-by-side comparison:

Cleaned Tokens: Words that are lowercased and filtered (no punctuation or stop words) but still in their natural grammatical form.

Stemmed Tokens: The final "shaved" versions of those words.

Why do we do this? This process significantly reduces the "vocabulary size" of your dataset. By mapping different versions of a word to a single root, your search engine or AI model becomes much more efficient at finding relevant information, regardless of the tense or pluralization used.

```python
# Step 5: Apply stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

print(" STEP 5: Stemmed Tokens\n")
for i in range(0, len(stemmed_tokens), 10):
    print(' '.join(stemmed_tokens[i:i+10]))

print("\n" + "="*60)
print(" FINAL OUTPUT")
print("="*60 + "\n")

print(" Tokens (Cleaned & Filtered):")
for i in range(0, len(filtered_tokens), 5):
    print(', '.join(filtered_tokens[i:i+5]))

print("\n" + "-"*60 + "\n")

print("Stemmed (Final Roots):")
for i in range(0, len(stemmed_tokens), 5):
    print(', '.join(stemmed_tokens[i:i+5]))
```
